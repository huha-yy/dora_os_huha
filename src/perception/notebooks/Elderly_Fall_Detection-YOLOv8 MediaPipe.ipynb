## Person Pose Estimation

```python
POSE_CONNECTIONS = [
    (11, 13), (13, 15),
    (12, 14), (14, 16),
    (11, 12),
    (23, 24),
    (11, 23), (12, 24),
    (23, 25), (25, 27),
    (24, 26), (26, 28),
]

def draw_pose(image, landmarks):
    h, w, _ = image.shape

    # Draw keypoints
    for lm in landmarks:
        x = int(lm.x * w)
        y = int(lm.y * h)
        cv2.circle(image, (x, y), 3, (0, 255, 0), -1)

    # Draw skeleton
    for a, b in POSE_CONNECTIONS:
        x1 = int(landmarks[a].x * w)
        y1 = int(landmarks[a].y * h)
        x2 = int(landmarks[b].x * w)
        y2 = int(landmarks[b].y * h)
        cv2.line(image, (x1, y1), (x2, y2), (255, 0, 0), 2)
```

```python
base_options = python.BaseOptions(
    model_asset_path="/home/frank/models/pose_landmarker_lite.task"
)

options = vision.PoseLandmarkerOptions(
    base_options=base_options,
    running_mode=vision.RunningMode.IMAGE,
    min_pose_detection_confidence=0.5,
    min_pose_presence_confidence=0.5,
    min_tracking_confidence=0.5,
)

pose_landmarker = vision.PoseLandmarker.create_from_options(options)
```

```python
cap = cv2.VideoCapture(video_path)

out = cv2.VideoWriter(
    "/home/frank/data/fall_detection/2_out_video_pose.mp4",
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (width, height),
)

# -------------------------
# Main loop (YOLO → Pose)
# -------------------------
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)  # your YOLO model

    for result in results:
        for bbox, cls in zip(result.boxes.xyxy, result.boxes.cls):
            if int(cls) != 0:  # person
                continue

            x1, y1, x2, y2 = map(int, bbox)
            person_bbox = frame[y1:y2, x1:x2]

            if person_bbox.size == 0:
                continue

            # BGR → RGB
            rgb = cv2.cvtColor(person_bbox, cv2.COLOR_BGR2RGB)

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb,
            )

            detection_result = pose_landmarker.detect(mp_image)

            if detection_result.pose_landmarks:
                landmarks = detection_result.pose_landmarks[0]
                draw_pose(person_bbox, landmarks)

            frame[y1:y2, x1:x2] = person_bbox
            cv2.rectangle(frame, (x1, y1), (x2, y2), (255, 0, 0), 2)

    out.write(frame)

# -------------------------
# Cleanup
# -------------------------
cap.release()
out.release()
pose_landmarker.close()
```

## Standing or Lying Down

```python
def calculate_angle(shoulder_center, hip_center):
    dy = shoulder_center[1] - hip_center[1]
    dx = shoulder_center[0] - hip_center[0]
    angle = math.atan2(dy, dx)
    return abs(90 - np.degrees(angle))

def classify_posture(torso_angle, standing_threshold=10, lying_threshold=60):
    if torso_angle < standing_threshold:
        return "Standing"
    elif torso_angle > lying_threshold:
        return "Lying Down"
    else:
        return "Falling"
```

```python
cap = cv2.VideoCapture(video_path)

out = cv2.VideoWriter(
    "/home/frank/data/fall_detection/3_out_video_act.mp4",
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (width, height),
)

pose_results = []

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)  # your YOLO model

    for result in results:
        for bbox, cls in zip(result.boxes.xyxy, result.boxes.cls):
            if int(cls) != 0:  # person
                continue

            x1, y1, x2, y2 = map(int, bbox)
            person_bbox = frame[y1:y2, x1:x2]

            if person_bbox.size == 0:
                continue

            # BGR → RGB
            rgb = cv2.cvtColor(person_bbox, cv2.COLOR_BGR2RGB)

            mp_image = mp.Image(
                image_format=mp.ImageFormat.SRGB,
                data=rgb,
            )

            detection_result = pose_landmarker.detect(mp_image)
            d = {}
            if detection_result.pose_landmarks:
                landmarks = detection_result.pose_landmarks[0]
                draw_pose(person_bbox, landmarks)

                h, w, _ = person_bbox.shape

                # Shoulders
                left_shoulder  = landmarks[11]
                right_shoulder = landmarks[12]
                
                shoulders = [
                    (left_shoulder.x * w, left_shoulder.y * h),
                    (right_shoulder.x * w, right_shoulder.y * h),
                ]

                # Hips
                left_hip  = landmarks[23]
                right_hip = landmarks[24]

                hips = [
                    (left_hip.x * w, left_hip.y * h),
                    (right_hip.x * w, right_hip.y * h),
                ]

                shoulder_center = (
                    (shoulders[0][0] + shoulders[1][0]) / 2,
                    (shoulders[0][1] + shoulders[1][1]) / 2,
                )

                hip_center = (
                    (hips[0][0] + hips[1][0]) / 2,
                    (hips[0][1] + hips[1][1]) / 2,
                )

                torso_angle = calculate_angle(shoulder_center, hip_center)
                posture = classify_posture(torso_angle)
                d.update({
                    'left_shoulder':left_shoulder,
                    'right_shoulder': right_shoulder,
                    'left_hip': left_hip,
                    'right_hip': right_hip,
                    'shoulders': shoulders,
                    'hips':hips,
                    'shoulder_center': shoulder_center,
                    'hip_center': hip_center,
                    'torso_angle':torso_angle,
                    'posture':posture,
                })
                
                # Overlay text on full frame
                cv2.putText(
                    frame,
                    f"Posture: {posture}",
                    (x1, y1 - 30),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    1,
                    (255, 255, 255),
                    2,
                    cv2.LINE_AA,
                )

                cv2.putText(
                    frame,
                    f"Angle: {torso_angle:.2f} deg",
                    (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    1,
                    (255, 255, 255),
                    2,
                    cv2.LINE_AA,
                )

                pose_results.append(d)


            frame[y1:y2, x1:x2] = person_bbox
            cv2.rectangle(frame, (x1, y1), (x2, y2), (255, 0, 0), 2)
            pose_results.append(d)

    out.write(frame)

# -------------------------
# Cleanup
# -------------------------
cap.release()
out.release()
pose_landmarker.close()
```
